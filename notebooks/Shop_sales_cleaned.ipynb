```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

```python
#Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("%", "percent")
)

df.columns
```

```python
#Clean text Columns
text_columns = [
    "customername",
    "product",
    "category",
    "paymentmethod",
    "region"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
```

```python
#standardize Customer names
df["customername"] = df["customername"].str.title()
```

```python
#replace missing names with unknown
df["customername"] = df["customername"].fillna("Unknown")
```

```python
#standardiz product names
df["product"] = df["product"].str.title()
```

```python
sorted(df["product"].dropna().unique())
```

```python
#Product with suspicious spaces
df[df["product"].str.contains(r"\s{2,}", na=False)]["product"].unique()
```

```python
df["product"] = df["product"].str.replace(r"\s+", " ", regex=True)
```

```python
df["category"].value_counts(dropna=False)
```

```python
df["category"] = df["category"].str.strip().str.title()
```

```python
df["paymentmethod"].value_counts(dropna=False)
```

```python
payment_mapping = {
    "cash": "Cash",
    "card": "Card",
    "visa/card": "Card",
    "mpesa": "M-Pesa",
    "m-pesa": "M-Pesa"
}

df["paymentmethod"] = (
    df["paymentmethod"]
    .str.lower()
    .str.strip()
    .replace(payment_mapping)
)
```

```python
df["paymentmethod"].value_counts(dropna=False)
```

```python
df["paymentmethod"] = df["paymentmethod"].fillna("Unknown")
```

```python
df["region"] = df["region"].str.strip().str.title()
```

```python
df["region"].value_counts(dropna=False)
```

```python
df["region"] = df["region"].fillna("Unknown")
```

```python
#convert date to time
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)
```

```python
#Date range
print("Earliest date:", df["date"].min())
print("Latest date:", df["date"].max())
```

```python
df["sales_amount"] = df["quantity"] * df["unitprice"]
```
